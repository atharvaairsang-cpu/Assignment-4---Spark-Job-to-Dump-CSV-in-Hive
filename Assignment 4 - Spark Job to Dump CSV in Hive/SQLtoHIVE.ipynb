```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("HiveTest") \
    .master("spark://spark-master:7077") \
    .enableHiveSupport() \
    .getOrCreate()
```

```python
spark.sql("DROP TABLE IF EXISTS beeline_test")
spark.sql("CREATE TABLE beeline_test (id INT, msg STRING)")
spark.sql("INSERT INTO beeline_test VALUES (1, 'created from pyspark')")
spark.sql("SELECT * FROM beeline_test").show()
```
